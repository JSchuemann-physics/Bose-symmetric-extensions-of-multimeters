```python
import numpy as np
```

```python
ATOL = 1E-8
```

```python
def is_herm(matrix: np.ndarray) -> bool:
    """Check whether `matrix` is Hermitian."""

    return np.allclose(matrix, matrix.conj().T)

def is_psd(matrix: np.ndarray) -> bool:
    """Test `matrix` for positive semi-definiteness."""

    if is_herm(matrix):
        return np.all(np.linalg.eigvalsh(matrix) >= - ATOL)
    else:
        return False
```

```python
eta = 3/5
zero_psi = np.sqrt(5/6) 
one_psi = np.sqrt(1/6)

assert zero_psi**2 + one_psi**2 == 1 #check for normalization

A = np.zeros((5,5))
#diagonals
A[0][0] = 2/5 * one_psi**2
A[1][1] = 2/5 * one_psi**2
A[2][2] = 1/5
A[3][3] = 2/5 * zero_psi**2
A[4][4] = 2/5 * zero_psi**2

#off-diagonals
A[0][1] = 1/2 * eta * zero_psi * one_psi
A[1][0] = 1/2 * eta * zero_psi * one_psi

A[2][1] = 1/np.sqrt(6) * eta * zero_psi * one_psi
A[1][2] = 1/np.sqrt(6) * eta * zero_psi * one_psi

A[3][2] = 1/np.sqrt(6) * eta * zero_psi * one_psi
A[2][3] = 1/np.sqrt(6) * eta * zero_psi * one_psi

A[4][3] = 1/2 * eta * zero_psi * one_psi
A[3][4] = 1/2 * eta * zero_psi * one_psi


print(A)
print(is_psd(A))
```

```python
np.linalg.eigvalsh(A)
```
