```python
import itertools

def compositions(k, d):
    """
    Generate all vectors in N^d whose entries sum to k. 
    See https://en.wikipedia.org/wiki/Stars_and_bars_(combinatorics)
    """
    for dividers in combinations(range(k + d - 1), d - 1):
        prev = -1
        result = []
        for div in dividers + (k + d - 1,):
            result.append(div - prev - 1)
            prev = div
        yield tuple(result)

#use math.comb instead
def binom(n, k):
        if n < k:
            return 0
        else:
            return comb(n, k)

def dim_sym(k, d):
    return comb(k + d - 1, d - 1) 
```

```python
def unrank(n, k):
    """
    Computes the combinations of (ck-1,..., c1, c0) that corresponds to n in the Combinatorial Number system. 
    See https://en.wikipedia.org/wiki/Combinatorial_number_system.
    """
    c = []
    k_first = k
    c_i = k-1
    c_prev = c_i
    for i in range(k):
        while comb(c_i, k) <= n:
            c_prev = c_i
            c_i += 1
        c.append(c_prev)

        #check if we have already found all summands (could be commented out)
        #if comb(c_prev, k) == n:
        #    c[i+1:] = reversed(range(k_first-i-1))
        #    break
            
        #prepare the next iteration
        n -= comb(c_prev, k)
        k -= 1
        c_i = k-1

    return tuple(reversed(c))
```

```python
def rank(c):
    """
    Computes the rank of a combinations (c0, ..., ck) in the combinatorial number system.
    """
    rank = 0
    for i, c_i in enumerate(c):
        rank += comb(c_i, i+1)
    return rank
```

### Qutrit sums

```python
def sum1(n:int, pr:bool):
    k = 5 + n * 6
    a = []
    for w0 in range((k+1)//3, (k-1)//2+1):
        for w1 in range(k-2*w0 + 1, w0+1):
            w2 = k - w0 - w1
            a.append(tuple([w0, w1, w2]))
            if pr:
                print(tuple([w0, w1, w2]))
                print("______")
    return a

def sum2(n:int, pr:bool):
    a = []
    k = 5 + n * 6
    for w0 in range((k+1)//2, k+1):
        for w1 in range(0, k-w0+1):
            w2 = k - w0 - w1
            a.append(tuple([w0, w1, w2]))
            if pr:
                print(tuple([w0, w1, w2]))
                print("______")
    
```

```python
def sum1(n:int, pr:bool):
    k = 3 + 4 * n
    a = []
    for w0 in range((k+1)//4, (k-1)//2 +1):
        for w3 in range((k-3)//4, w0-1 +1):
            for w1 in range(k-2*w0-w3, w0 +1):
                w2 = k - w0 - w1 - w3
                a.append(tuple([w0, w1, w2, w3]))
                if pr:
                    print(tuple([w0, w1, w2, w3]))
                    print("______")
    return a
```

```python
def filter_comp(comps):
    res = []
    d = len(comps[0])
    for tup in comps:
        w0 = tup[0]
        belongs = True
        if w0 <= tup[1]:
            belongs = False
        for i in range(1, d-1):
            if w0 < tup[i+1]:
                belongs = False
        if belongs:
            res.append(tup)
    return res
```
