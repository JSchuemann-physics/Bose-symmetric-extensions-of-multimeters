```python
def is_herm(matrix: np.ndarray) -> bool:
    """Check whether `matrix` is Hermitian."""

    return np.allclose(matrix, matrix.conj().T)


def is_psd(matrix: np.ndarray) -> bool:
    """Test `matrix` for positive semi-definiteness."""

    if is_herm(matrix):
        return np.all(np.linalg.eigvalsh(matrix) >= - ATOL)
    else:
        return False

def is_measurement(meas: list[np.ndarray]) -> bool:
    """Check whether `meas` is a well-defined quantum measurement.

    Args:
        meas (list): list of ndarrays representing the measurement's effects.

    Returns:
        bool: returns `True` iff `meas` is composed of effects which are:
            - Square matrices with the same dimension,
            - Positive semi-definite, and
            - Sum to the identity operator.
    """

    dims = meas[0].shape

    try:
        square = len(dims) == 2 and dims[0] == dims[1]
        same_dim = np.all(np.asarray([eff.shape for eff in meas]) == dims)
        psd = np.all([is_psd(eff) for eff in meas])
        complete = np.allclose(sum(meas), np.eye(dims[0]))
    except (ValueError, np.linalg.LinAlgError):
        return False

    return square and same_dim and psd and complete
```

```python
def rank(c):
    """
    Computes the rank of a combinations (c0, ..., ck) in the combinatorial number system.
    """
    rank = 0
    for i, c_i in enumerate(c): 
        rank += comb(c_i, i+1)
    return rank
#calling this function everytime feels soooo inefficient (think about it later)
def get_Eij(i, j, dim):
    M = np.zeros((dim, dim))
    M[i][j] = 1
    return M
```

```python
def sym_partial_trace(A: pc.HermitianVariable, dim: int, k: int):
    """Computes the partial trace of A over symmetric subspace.

    Args:
        A (pc.HermitianVariable): Hermitian Operator on symmetric subspace.
        d (int): dimension of the underlying Hilbert Space.
        k (int): number of marginal systems

    Returns: 
        (pc.HermitianVariable): (d x d)-matrix that represents the partial trace of A. 
    """
    total_pos = k+dim-1
    terms = []
    #E = [[pc.Constant(f"|{l}><{k}|", get_Eij(l, k, dim)) for k in range(dim)] for l in range(dim)]
    E = [[get_Eij(l, k, dim) for k in range(dim)] for l in range(dim)]

    for c in combinations(range(total_pos), dim-1):
        
        ds = rank(c) #dicke state index
        A_ds = A[ds,ds]
        w = [c[0]] + [c[n]-c[n-1]-1 for n in range(1, dim-1)] + [total_pos-1-c[dim-2]] #notice + is append here
        for i in range(0, dim):
            terms.append(w[i] * A_ds * E[i][i])
            for j in range(1+i, dim):                         
                #the terms where w[i] = 0 are zero but we have to explicitly check, otherwise
                #we get an indexing error for A[ds, ds_lshift]
                if w[i] > 0: 
                    c_lshift = np.array(c)
                    c_lshift[i:j] -= 1
                    ds_lshift = rank(c_lshift)
                    terms.append(np.sqrt(w[i]*(w[j]+1)) * A[ds, ds_lshift] * E[i][j])
                if w[j] > 0:
                    c_rshift = np.array(c)
                    c_rshift[i:j] += 1
                    ds_rshift = rank(c_rshift)
                    terms.append(np.sqrt(w[j]*(w[i]+1)) * A[ds, ds_rshift] * E[j][i])
    return sum(terms) / k #pc.sum produces a weird string, sum(terms) is more readable
```

```python
def get_extensionSDP(meas: list[np.ndarray], k: int):
    """
    Args:
        meas (list): list of ndarrays represetening the measurement's effects.
        k (int): level of the extension
    Returns:
        pc.Problem(): SDP for finding the amount of noise need for meas to have a 
        Bose-Symmetric extension to level k. 
    """
    assert(is_measurement(meas))

    dim = meas[0].shape[0] #dimension of the Hilbert space
    dim_sym = comb(k + dim - 1, dim - 1) #dimension of symmetric subspace Sym(H, k)
    n_a = len(meas) #number of effects

    SDP = pc.Problem()

    # Constants
    noise = pc.Constant("noise", np.eye(dim) / dim)
    sym_proj = pc.Constant("sym_proj", dim * np.eye(dim_sym) / dim_sym)
    M = [pc.Constant(f"M_{a}", meas[a]) for a in range(n_a)]

    # Variables
    eta = pc.RealVariable("eta", 1)
    SDP.add_constraint(eta <= 1)

    N = [pc.HermitianVariable(f"N_{a}", dim_sym) for a in range(n_a)]

    SDP.add_list_of_constraints([n >> 0 for n in N])
    SDP.add_constraint(pc.sum([n for n in N]) == sym_proj)

    for a in range(n_a):
        SDP.add_constraint(eta * M[a] + (1-eta) * pc.trace(M[a]) * noise == sym_partial_trace(N[a],dim, k))
    SDP.set_objective("max", eta)

    return SDP
```

```python
##flagged SIC-POVM (claude code)
# ─────────────────────────────────────────────
# Helper: d=3 Weyl-Heisenberg operators
# ─────────────────────────────────────────────
d = 3
omega = np.exp(2j * np.pi / d)          # primitive d-th root of unity
tau   = -np.exp(-1j * np.pi / d)        # phase factor τ = -e^{-iπ/d}

# Shift operator X: X|j⟩ = |j⊕1 mod d⟩
X3 = np.roll(np.eye(d, dtype=complex), shift=1, axis=0)

# Clock operator Z: Z|j⟩ = ω^j |j⟩
Z3 = np.diag([omega**j for j in range(d)])

def weyl_heisenberg(i1, i2, X, Z, tau):
    """Displacement operator D_{i1,i2} = τ^{i1·i2} X^{i1} Z^{i2}"""
    Xi = np.linalg.matrix_power(X, i1)
    Zi = np.linalg.matrix_power(Z, i2)
    return tau**(i1 * i2) * (Xi @ Zi)


# ─────────────────────────────────────────────
# 1. d=3 Hesse SIC-POVM  (SIC3c, φ=0)
#    fiducial vector |ψ₀⟩ = 1/√2 (|1⟩ − |2⟩)   [Eq. 31, φ=0]
#    effects  Mᵢ = (1/d) Dᵢ |ψ₀⟩⟨ψ₀| Dᵢ†
# ─────────────────────────────────────────────
psi0_3d = np.array([0, 1, -1], dtype=complex) / np.sqrt(2)
M0_3d   = np.outer(psi0_3d, psi0_3d.conj()) / d   # rank-1 seed, trace = 1/d

M_hesse = []
for i1 in range(d):
    for i2 in range(d):
        D = weyl_heisenberg(i1, i2, X3, Z3, tau)
        M_hesse.append(D @ M0_3d @ D.conj().T)

# Sanity checks
total_3d = sum(M_hesse)
print("=== d=3 Hesse SIC-POVM ===")
print(f"Number of effects : {len(M_hesse)}  (expected 9 = d²)")
print(f"Sum = I₃?          {np.allclose(total_3d, np.eye(d))}")
print(f"Each effect PSD?   {all(np.all(np.linalg.eigvalsh(M) >= -1e-10) for M in M_hesse)}")


# ─────────────────────────────────────────────
# 2. Flagged SIC-POVM in d=4
#    • Embed each 3×3 Hesse-SIC effect into 4×4 by zero-padding
#    • Append |3⟩⟨3| as the flag effect
#    Total: 9 + 1 = 10 effects, acting on ℂ⁴
# ─────────────────────────────────────────────
D4 = 4

def embed_3in4(M3):
    """Pad a 3×3 matrix into the top-left corner of a 4×4 zero matrix."""
    M4 = np.zeros((D4, D4), dtype=complex)
    M4[:3, :3] = M3
    return M4

M_flagged = [embed_3in4(M) for M in M_hesse]

# Flag effect: |3⟩⟨3|
flag_effect = np.zeros((D4, D4), dtype=complex)
flag_effect[3, 3] = 1.0
M_flagged.append(flag_effect)

## Sanity checks
#total_4d = sum(M_flagged)
#print("\n=== Flagged Hesse SIC-POVM in d=4 ===")
#print(f"Number of effects : {len(M_flagged)}  (expected 10 = d² + 1)")
#print(f"Sum = I₄?          {np.allclose(total_4d, np.eye(D4))}")
#print(f"Each effect PSD?   {all(np.all(np.linalg.eigvalsh(M) >= -1e-10) for M in M_flagged)}")
#
## Display effects
#for i, M in enumerate(M_flagged):
#    label = f"flag |3><3|" if i == 9 else f"M[{i}]  (i1={i//d}, i2={i%d})"
#    print(f"\n{label}:")
#    print(np.round(M, 6))
```

```python
k = 10
SDP = get_extensionSDP(M_flagged, k)
#print(SDP)
```
