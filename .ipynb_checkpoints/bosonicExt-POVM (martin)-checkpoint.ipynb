```python
def is_herm(matrix: np.ndarray) -> bool:
    """Check whether `matrix` is Hermitian."""

    return np.allclose(matrix, matrix.conj().T)


def is_psd(matrix: np.ndarray) -> bool:
    """Test `matrix` for positive semi-definiteness."""

    if is_herm(matrix):
        return np.all(np.linalg.eigvalsh(matrix) >= - ATOL)
    else:
        return False

def is_measurement(meas: list[np.ndarray]) -> bool:
    """Check whether `meas` is a well-defined quantum measurement.

    Args:
        meas (list): list of ndarrays representing the measurement's effects.

    Returns:
        bool: returns `True` iff `meas` is composed of effects which are:
            - Square matrices with the same dimension,
            - Positive semi-definite, and
            - Sum to the identity operator.
    """

    dims = meas[0].shape

    try:
        square = len(dims) == 2 and dims[0] == dims[1]
        same_dim = np.all(np.asarray([eff.shape for eff in meas]) == dims)
        psd = np.all([is_psd(eff) for eff in meas])
        complete = np.allclose(sum(meas), np.eye(dims[0]))
    except (ValueError, np.linalg.LinAlgError):
        return False

    return square and same_dim and psd and complete
```

```python
def get_extensionSDP(meas: list[np.ndarray], k: int):

    assert(is_measurement(meas))

    dim = meas[0].shape[0]
    assert(dim == 2)
    n_a = len(meas)

    SDP = pc.Problem()

    # Constants
    noise = pc.Constant("noise", np.eye(dim) / dim)
    sym_proj = pc.Constant("sym_proj", 2 * np.eye(k + 1) / (k + 1))
    M = [pc.Constant(f"M_{a}", meas[a]) for a in range(n_a)]

    E00 = pc.Constant("|0><0|", np.array([[1, 0], [0, 0]]))
    E01 = pc.Constant("|0><1|", np.array([[0, 1], [0, 0]]))
    E10 = pc.Constant("|1><0|", np.array([[0, 0], [1, 0]]))
    E11 = pc.Constant("|1><1|", np.array([[0, 0], [0, 1]]))

    # Variables
    vis = pc.RealVariable("vis", 1)
    SDP.add_constraint(vis <= 1)

    N = [pc.HermitianVariable(f"N_{a}", k+1) for a in range(n_a)]

    SDP.add_list_of_constraints([n >> 0 for n in N])
    SDP.add_constraint(pc.sum([n for n in N]) == sym_proj)

    for a in range(n_a):
        SDP.add_constraint(vis * M[a] + (1-vis) * pc.trace(M[a]) * noise == N[a][0,0]*E00 + pc.sum([N[a][i,i] * ((k-i)*E00/k + i*E11/k) + (N[a][i-1,i]*E01 + N[a][i,i-1]*E10) * np.sqrt(i * (k-i+1)) / k for i in range(1, k + 1)]))

    SDP.set_objective("max", vis)

    return SDP
```

```python
X = np.array([[0, 1],
              [1, 0]])
Z = np.array([[1, 0],
              [0, -1]])

def get_projector(phi):
    proj = 0.5*(np.eye(2) + np.cos(phi) * X + np.sin(phi) * Z)
    proj.real[abs(proj.real) < ATOL] = 0.0
    proj.real[abs(proj.real) < ATOL] = 0.0
    return proj
```

Projective measurement in $|+\rangle, |-\rangle$ basis

```python
meas = [get_projector(0), get_projector(np.pi)]
```

Projective measurement in $|0\rangle, |1\rangle$ basis

```python
meas = [get_projector(np.pi/2), get_projector(3*np.pi/2)]
```

POVM in $XZ$-plane

```python
meas = [(2/3) * get_projector(np.pi/2), (2/3) * get_projector(7*np.pi/6), (2/3) * get_projector(11*np.pi/6)]
```

SIC-POVM

```python
meas = [0.5 * np.array([[1, 0],
                        [0, 0]]),
        0.5 * np.array([[1/3, np.sqrt(2)/3],
                        [np.sqrt(2)/3, 2/3]]),
        0.5 * np.array([[1/3, np.exp(1j*2*np.pi/3)*np.sqrt(2)/3],
                        [np.exp(-1j*2*np.pi/3)*np.sqrt(2)/3, 2/3]]),
        0.5 * np.array([[1/3, np.exp(1j*4*np.pi/3)*np.sqrt(2)/3],
                [np.exp(-1j*4*np.pi/3)*np.sqrt(2)/3, 2/3]])]
```

```python
for m in meas:
    print(m)
```

```python
for a in range(len(meas)):
    for b in range(len(meas)):
        trace = np.real_if_close(np.trace(meas[a] @ meas[b]))
        print(f"Tr(M_{a} M_{b})) = {trace}")
```

```python
# get_extensionSDP(measurement, k)
SDP = get_extensionSDP(meas, 30)
print(SDP)
```

```python
start_time = time.time()
SDP.solve(solver="mosek", verbosity=1)
# SDP.solve(solver="cvxopt", verbosity=1)
end_time = time.time()
```

```python
k_max = 15

for k in range(5, k_max + 1):
    print(f"k = {k}")

    start_time = time.time()
    SDP = get_extensionSDP(meas, k)
    SDP.solve(solver="mosek", verbosity=0)
    # SDP.solve(solver="cvxopt", verbosity=0)
    end_time = time.time()

    vis = SDP.get_variable("vis").value

    print()
    print(f"vis = {vis}")
    print(f"runtime: {end_time - start_time} seconds")
    print()
```
